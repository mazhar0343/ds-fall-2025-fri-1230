```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("/workspaces/ds-fall-2025-fri-1230/Week-03-EDA-and-Dashboards/data/movie_ratings.csv")
sns.set()
```

Question 1
What's the breakdown of genres for the movies that were rated?

```python
genre_counts = df["genres"].value_counts()
genre_counts.plot(kind="barh")
```

Question 2
 Which genres have the highest viewer satisfaction (highest ratings)?

```python
average_ratings = df.groupby("genres")["rating"].mean()
average_ratings.plot(kind="barh")
```

Quesiton 3
How does mean rating change across movie release years?

```python
ratings_yearly = df.groupby("year")["rating"].mean()
ratings_yearly.plot(kind="line")
```

Question 4
What are the 5 best-rated movies that have at least 50 ratings? At least 150 ratings?

```python
movie_stats = df.groupby('title').agg(
    mean_rating=('rating', 'mean'),
    rating_count=('rating', 'count')
)
top5_50 = movie_stats[movie_stats['rating_count'] >= 50].sort_values(by='mean_rating', ascending=False).head(5)
top5_150 = movie_stats[movie_stats['rating_count'] >= 150].sort_values(by='mean_rating', ascending=False).head(5)
print(top5_50, top5_150)
```
